```python
cv = CountVectorizer(max_features = 100, ngram_range = (1, 2))
```

```python
sentence = "The Eiffel Tower, designed by Gustave Eiffel and built for the 1889 World's Fair (Exposition Universelle), was completed in record time, taking only two years, two months, and five days to construct, from 1887 to 1889."
```

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stopwords_list = stopwords.words('english')
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
corpus = [];
for i in range(0, len(sentence)):
    words = sentence.split()
    words = [ps.stem(word) for word in words if word not in set(stopwords_list)]
    words = ' '.join(words)
    corpus.append(words)
```

```python
cv.fit_transform(corpus).toarray()
```
