### Parts of speech tagging

```python
sentence = "The Eiffel Tower, designed by Gustave Eiffel and built for the 1889 World's Fair (Exposition Universelle), was completed in record time, taking only two years, two months, and five days to construct, from 1887 to 1889."
```

```python
import nltk
words = nltk.word_tokenize(sentence)
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
tag_elements = nltk.pos_tag(words)
```

### Named Entity Recognition

```python
nltk.download('maxent_ne_chunker_tab')
```

```python
nltk.download('words')
```

```python
nltk.download('maxent_ne_chunker')
```

```python
nltk.ne_chunk(tag_elements)
```

```python
nltk.ne_chunk(tag_elements).draw()
```
